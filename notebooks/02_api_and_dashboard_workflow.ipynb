# API and Dashboard Workflow
The dashboard talks to the FastAPI backend over HTTP.

```python
import httpx

base_url = 'http://127.0.0.1:8000'
with httpx.Client(base_url=base_url, timeout=10.0) as client:
    print(client.get('/health').json())
    print(client.get('/state').json())
```

```python
# Start the backend with: ./venv/bin/python main.py --serve
# Start the dashboard with: docker compose up -d grafana
print('API-driven dashboard workflow documented in the notebook.')
```
